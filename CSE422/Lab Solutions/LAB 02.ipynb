```python
import random

def double_point_crossover(p1, p2):

  # Check the lengths. We cannot perform crossover if lengths of parents are not equal.
  if len(p1)== len(p2):
    idx1= random.randint(1, 10)
    idx2= random.randint(1, 10)
  # Choose two random crossover indices.
  # random.randint(1, 10) -> generates a random integer between 1 and 10

  # Perform double-point crossover to produce two children.
    if idx1>=idx2:
      child1= p1[idx2:idx1]
      child2= p2[idx2:idx1]

      finalchild1= p1[:idx2] + child2 + p1[idx1:len(p1)]
      finalchild2= p2[:idx2] + child1 + p2[idx1:len(p2)]

    else:
      child1 = p1[idx1:idx2]
      child2 = p2[idx1:idx2]

      finalchild1 = p1[:idx1] + child2 +p1[idx2:len(p1)]
      finalchild2 =  p2[:idx1] + child1 +p2[idx2:len(p2)]
  # arr[m:n] -> extracts element from indices m to (n-1) from arr
  # arr[:n] -> extracts element from indices 0 to (n-1) from arr


  # Return the children
  else:
    print(-1)

  return finalchild1, finalchild2
```
