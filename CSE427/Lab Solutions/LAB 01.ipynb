```python
import numpy as np
import random
```

```python
#task1
arr1= np.zeros(20, dtype="int")
arr2= np.zeros(20, dtype="int")
for i in range(20):
  arr1[i]=random.randint(1, 100)
  arr2[i]=random.randint(1, 100)
print(arr1)
print(arr2)
```

```python
#task2
mat1= arr1.reshape(5,4)
mat2= arr2.reshape(5,4)

print(mat1)
print()
print(mat2)
```

```python
#task 3
print(mat1 * mat2)
```

```python
#task 4
matb = np.array([[ 7 , 4 , 8 , 5],
 [ 7, 10 , 3  ,7],
 [ 8 , 5,  4,  8],
 [ 8 , 3 , 6 , 5],
 [ 2 , 8 , 6,  2]])

matc= np.array(mat1 * matb)
print(matc)
```

```python
#task 5
new =np.array(matc[0:5, 1:3])
print(new)
```

```python
#task 6
print(np.max(new))
row, column =np.unravel_index(np.argmax(new), new.shape)
print(row,column)
```

```python
#task 7
newest = new.flatten()
print(newest)
```

Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
#task 1
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]

plt.title("temperature_dhaka vs humidity_dhaka")
plt.xlabel("values of temperature_dhaka")
plt.ylabel("values of humidity_dhaka")
plt.grid()
plt.scatter(temperature_dhaka, humidity_dhaka, color = "red", marker = "*")
plt.show()
```

```python
#task 2
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
plt.figure(figsize=(12, 8))

plt.plot(study_hours, marks)
plt.show()
```

```python
#task 3
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
plt.barh(subjects, marks)
plt.show()
```

```python
#task 4
colors = ['red', 'green', 'blue', 'violet', 'black']
plt.bar(subjects, marks, color=colors)
plt.show()
```

```python
#task 5
plt.figure(figsize=(12, 8))

#for 1
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, color = "red", marker = "*")

#for 2
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
plt.subplot(2, 2, 2)
plt.plot(study_hours, marks)

#for 3
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
plt.subplot(2, 2, 3)
plt.barh(subjects, marks)


#for 4
colors = ['red', 'green', 'blue', 'violet', 'black']
plt.subplot(2, 2, 4)
plt.bar(subjects, marks, color=colors)

plt.show()
```
